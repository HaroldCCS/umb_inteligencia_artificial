```python
# ================================
# IA - Algoritmos de búsqueda en Grafos
# Caso de turismo
# ================================

import networkx as nx
import matplotlib.pyplot as plt
from queue import PriorityQueue
import time

# Crear el grafo con pesos (costos/distancias)
G = nx.Graph()
edges = [
    ("Hotel", "Museo", 2),
    ("Hotel", "Playa", 5),
    ("Museo", "Parque", 4),
    ("Playa", "Parque", 2),
    ("Parque", "Restaurante", 3),
    ("Museo", "Restaurante", 7)
]
G.add_weighted_edges_from(edges)

pos = nx.spring_layout(G, seed=42)

# Función para visualizar el grafo resaltando el camino recorrido
def draw_graph(path=[], visited=[]):
    plt.figure(figsize=(8,6))
    nx.draw(G, pos, with_labels=True, node_size=1500, node_color="lightblue", font_size=10, font_weight="bold")
    labels = nx.get_edge_attributes(G, 'weight')
    nx.draw_networkx_edge_labels(G, pos, edge_labels=labels)

    # Resaltar nodos visitados
    nx.draw_networkx_nodes(G, pos, nodelist=visited, node_color="orange")
    # Resaltar el camino final
    if path:
        edges_path = list(zip(path, path[1:]))
        nx.draw_networkx_edges(G, pos, edgelist=edges_path, width=3, edge_color="red")
        nx.draw_networkx_nodes(G, pos, nodelist=path, node_color="green")
    plt.show()

# ================================
# Algoritmos de búsqueda
# ================================

def bfs(start, goal):
    visited, queue = set(), [[start]]
    while queue:
        path = queue.pop(0)
        node = path[-1]
        if node not in visited:
            visited.add(node)
            draw_graph(path, visited)   # Visual paso a paso
            time.sleep(1)
            if node == goal:
                return path
            for neighbor in G.neighbors(node):
                new_path = list(path)
                new_path.append(neighbor)
                queue.append(new_path)
    return None

def dfs(start, goal):
    visited, stack = set(), [[start]]
    while stack:
        path = stack.pop()
        node = path[-1]
        if node not in visited:
            visited.add(node)
            draw_graph(path, visited)
            time.sleep(1)
            if node == goal:
                return path
            for neighbor in G.neighbors(node):
                new_path = list(path)
                new_path.append(neighbor)
                stack.append(new_path)
    return None

def ucs(start, goal):
    visited = set()
    queue = PriorityQueue()
    queue.put((0, [start]))
    while not queue.empty():
        cost, path = queue.get()
        node = path[-1]
        if node not in visited:
            visited.add(node)
            draw_graph(path, visited)
            time.sleep(1)
            if node == goal:
                return path, cost
            for neighbor in G.neighbors(node):
                new_path = list(path)
                new_path.append(neighbor)
                queue.put((cost + G[node][neighbor]['weight'], new_path))
    return None

def greedy(start, goal, heuristic):
    visited, queue = set(), PriorityQueue()
    queue.put((heuristic[start], [start]))
    while not queue.empty():
        h, path = queue.get()
        node = path[-1]
        if node not in visited:
            visited.add(node)
            draw_graph(path, visited)
            time.sleep(1)
            if node == goal:
                return path
            for neighbor in G.neighbors(node):
                new_path = list(path)
                new_path.append(neighbor)
                queue.put((heuristic[neighbor], new_path))
    return None

def astar(start, goal, heuristic):
    visited, queue = set(), PriorityQueue()
    queue.put((heuristic[start], 0, [start]))
    while not queue.empty():
        f, g, path = queue.get()
        node = path[-1]
        if node not in visited:
            visited.add(node)
            draw_graph(path, visited)
            time.sleep(1)
            if node == goal:
                return path, g
            for neighbor in G.neighbors(node):
                new_path = list(path)
                new_path.append(neighbor)
                g_new = g + G[node][neighbor]['weight']
                f_new = g_new + heuristic[neighbor]
                queue.put((f_new, g_new, new_path))
    return None

# ================================
# DEMO
# ================================

heuristic = {
    "Hotel": 7,
    "Museo": 6,
    "Playa": 2,
    "Parque": 1,
    "Restaurante": 0
}

print("=== BFS ===")
path_bfs = bfs("Hotel", "Restaurante")
print("Camino encontrado BFS:", path_bfs)

print("\n=== DFS ===")
path_dfs = dfs("Hotel", "Restaurante")
print("Camino encontrado DFS:", path_dfs)

print("\n=== UCS ===")
path_ucs, cost_ucs = ucs("Hotel", "Restaurante")
print("Camino encontrado UCS:", path_ucs, "Costo:", cost_ucs)
```
